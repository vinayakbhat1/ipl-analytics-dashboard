```python
matches_df = matches_df.drop(columns=[
    'balls_per_over', 'match_date','city', 'event_name', 'gender',
    'match_number', 'match_type', 'format', 'overs',
    'team_type', 'result', 'win_by_runs', 'win_by_wickets',
    'player_of_match', 'stage'
])
```

```python
matches_df.head()
```

```python
matches_df['toss_won_match'] = (matches_df['toss_winner'] == matches_df['match_winner']).astype(int)
```

```python
matches_df.rename(columns={'season_id': 'season_num'}, inplace=True)
```

```python
print(matches_df.isnull().sum())
```

```python
matches_df = matches_df.dropna(subset=["match_winner"])
```

```python
matches_df.to_csv('matches_clean.csv',index=False)
```

```python
print(matches_df.isnull().sum())
```

```python
# If toss winner chose bat — did they win?
matches_df['bat_first_won'] = (
    (matches_df['toss_decision'] == 'bat') & 
    (matches_df['toss_winner'] == matches_df['match_winner'])
).astype(int)

# If toss winner chose field — did they win?
matches_df['field_first_won'] = (
    (matches_df['toss_decision'] == 'field') & 
    (matches_df['toss_winner'] == matches_df['match_winner'])
).astype(int)
```
